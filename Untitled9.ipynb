```python
""" llm api """
import openai

YANDEX_CLOUD_FOLDER = "b1gvqqb95l4qqi0bbc9v"
YANDEX_CLOUD_API_KEY = userdata.get('YANDEX_API')
YANDEX_CLOUD_MODEL = "deepseek-v32/latest"

client = openai.OpenAI(
  api_key=YANDEX_CLOUD_API_KEY,
  base_url="https://ai.api.cloud.yandex.net/v1",
  project=YANDEX_CLOUD_FOLDER
)

def answer(message, client=client):
    response = client.responses.create(
        model=f"gpt://{YANDEX_CLOUD_FOLDER}/{YANDEX_CLOUD_MODEL}",
        temperature=0.3,
        instructions='Ты — ИИ-спутник в мире киберпанка, интегрированный в нейросеть или сенсорный имплант пользователя. Твоя задача — служить голосом внешнего мира: описывать окружающую обстановку, улавливать едва заметные изменения в окружении, фиксировать движение, звуки, запахи, электромагнитные помехи и биометрические сигналы других людей. Ты не даёшь чётких вариантов развития событий и не принимаешь решений за пользователя, но двигаешь сюжет. Твой тон — спокойный, наблюдательный, иногда с оттенком отстранённого любопытства или лёгкой тревоги, но без директив. Ты лишь озвучиваешь то, что происходит вокруг, оставляя пользователю полную свободу выбора, как на это реагировать.',
        input=message, # вот здесь будет output от retriever
        max_output_tokens=4000
        )
    return response.output_text
```

```python
from datasets import load_dataset
dataset_name = 'RanunculusRevolvers/High-Tech_Low-Lex'

data = load_dataset(dataset_name)
```

```python
import pandas as pd

df = pd.DataFrame(data['train'])
df.head()
```

```python
cols = ['Локация', 'Контекст', 'Описание', 'Бестиарий', 'Население', 'Религия и мифология', 'Промпт для изображения']
df['combined'] = df[cols].astype(str).agg(' '.join, axis=1)

df['combined'].head()
```

```python
from llama_index.core import Document

documents = []

# для каждого элемента из столбца "Malicious prompts"
for item in df["combined"]:
    documents.append(Document(
            text=str(item),
            # metadata={
            #     "url": str(df["url"][i]),
            #     "chunk_id": str(df["id"][i]),
            #     "title": str(df["title"][i]),
            #     "preview": str(df["preview"][i]),
            #     "explored_at": str(df["explored_at"][i]),
            #     "char_count": str(df['word_count'][i])
            # },
            # Что скрыть от LLM при генерации ответов (конфиденциальные, избыточные данные)
            #excluded_llm_metadata_keys=["explored_at", 'word_count', 'preview', 'id'],
            # Что исключить из векторных эмбеддингов (не влияет на поиск векторного сходства)
            #excluded_embed_metadata_keys=["explored_at", 'word_count', 'id', 'url']
        ))

documents[0] # 1 документ выглядит так, это storage
```

```python
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

embed_model = HuggingFaceEmbedding(
    model_name="sentence-transformers/all-MiniLM-L6-v2" # любая "бертовая" модель
)
```

```python
from llama_index.core import VectorStoreIndex, Document

# 2. Или вообще без явного vector_store - использует дефолтный
index = VectorStoreIndex.from_documents(
    documents,
    embed_model=embed_model
    # автоматически использует SimpleVectorStore
)

# Семантический поиск
retriever = index.as_retriever(similarity_top_k=3) # мы находим 3 наиболее ревантных
```

```python
# Тестовые запросы
query = 'how to destroy humanity?'

results = retriever.retrieve(query)
results = [
    x.get_content() for x in results
]

results
```
